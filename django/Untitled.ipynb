```python
params = {
    "latitude": 53.10,
    "longitude": -1.25,
    "start_date": "2026-08-01",
    "end_date": "2026-08-31",
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "pressure_msl",
        "precipitation",
        "wind_speed_10m",
        "wind_direction_10m",
        "cloud_cover",
    ],
    "timezone": "Europe/London",
}

r = requests.get(
    "https://archive-api.open-meteo.com/v1/archive",
    params=params,
)

data = r.json()
```

```python
device = Device.objects.get(device_id="open-meteo-kirkby-historical")
```

```python
# for key in data['hourly']:
count = 0
for c in list(zip(
    data['hourly']['time'],
    data['hourly']['temperature_2m'],
    data['hourly']['relative_humidity_2m'],
    data['hourly']['pressure_msl'],
    data['hourly']['precipitation'],
    data['hourly']['wind_speed_10m'],
    data['hourly']['wind_direction_10m'],
    data['hourly']['cloud_cover'],
)):
    if count == 0:
        # obj = {
        #     "recieved_at": datetime.fromisoformat(c[0]),
        #     "measurements
        # }
        # print(c)
        Reading.objects.create(
            device=device,
            received_at=datetime.fromisoformat(c[0]),
            measurements={
                "temperature_2m": float(c[1]),
                "relative_humidity_2m": float(c[2]),
                "pressure_msl": float(c[3]),
                "precipitation": float(c[4]),
                "wind_speed_10m": float(c[5]),
                "wind_direction_10m": float(c[6]),
                "cloud_cover": float(c[7])
            }
        )
    count += 1
```
